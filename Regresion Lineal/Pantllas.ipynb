```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
datos = pd.DataFrame({
    "TiempoPantalla":[1.5, 3, 3.5, 4.5, 5, 6, 6.5, 7, 8.5, 9], 
    "Nivel_de_concentracion": [8, 7, 7.5, 6, 6.5, 5, 5.5, 4, 3, 3.5],
    "NivelCortisol":[6.2, 9.5, 8, 12.3, 11, 15.8, 14.1, 17.5, 22, 21.2]})

display(datos)
```

```python
modelo = smf.ols(
    formula="NivelCortisol ~ TiempoPantalla",
    data=datos
).fit()
```

```python
modelo.summary()
```

```python
modelo2 = smf.ols(
    formula="NivelCortisol ~ Nivel_de_concentracion + TiempoPantalla",
    data=datos
).fit()

modelo2.summary()
```
