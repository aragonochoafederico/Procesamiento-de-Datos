<h4>Cargar librerias<h4>

```python
import numpy as np                        # Algebra lineal
import pandas as pd                       # Manipular, limpiar y analizar datos
import statsmodels.formula.api as smf     # Modelos estadisticos
import matplotlib.pyplot as plt           # Graficar
```

<h4>Simular datos<h4>

```python
datos = pd.DataFrame({
    "area": [40,60,80,100,120,140],
    "precio": [165,305,315,375,385,525],
    "Estrato": [1,3,3,3,4,5]
})

display(datos)
```

<h4>Construir el modelo<h4>

```python
modelo = smf.ols(formula="precio ~ area", data=datos).fit()
```

<h4>Analizar el modelo<h4>

```python
# Gráfico de los datos
plt.scatter(datos["area"], datos["precio"], label="Datos")

# Recta de regresión
plt.plot(datos["area"], modelo.predict(datos), color="red", label="Modelo")

plt.xlabel("Área")
plt.ylabel("Precio")
plt.title("Regresión lineal: Precio vs Área")
plt.legend()

plt.show()
```

```python
modelo.summary()
```

```python
print("Coeficientes:", modelo.params)
print("R²:", modelo.rsquared)
print("MSE:", modelo.mse_resid)
print("Errores estándar:", modelo.bse)
print("Estadísticos t:", modelo.tvalues)
print("P-valores:", modelo.pvalues)
print("F:", modelo.fvalue)
print("P-valor F:", modelo.f_pvalue)
```

<h4>Prediciones<h4>

```python
modelo.predict({"area": 700})
```
